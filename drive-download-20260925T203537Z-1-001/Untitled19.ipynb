```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
import ipywidgets as widgets
from ipywidgets import interact

# Muestreo base invariante Z ~ N(0, I)
np.random.seed(42)
N = 3000
Z = np.random.multivariate_normal(mean=[0, 0], cov=np.eye(2), size=N)

def plot_bivariate_covariance(sigma_11=1.0, sigma_22=8.0, sigma_12=0.0):
    # Definición de la matriz de covarianza
    Sigma = np.array([[sigma_11, sigma_12],
                      [sigma_12, sigma_22]])

    # Verificación de definición positiva
    det = np.linalg.det(Sigma)
    if det <= 0:
        plt.figure(figsize=(6, 6))
        plt.text(0.5, 0.5, f"Matriz no definida positiva\n" + fr"$\det(\Sigma) = {det:.2f} \le 0$",
                 ha='center', va='center', fontsize=14, color='red')
        plt.xlim(0, 1)
        plt.ylim(0, 1)
        plt.axis('off')
        plt.show()
        return

    # Transformación lineal X = Z @ L^T donde Sigma = L L^T
    L = np.linalg.cholesky(Sigma)
    X = Z @ L.T

    # Estimación de densidad espectral para coloreado
    kde = gaussian_kde(X.T)
    density = kde(X.T)
    idx = density.argsort()
    x, y, density = X[idx, 0], X[idx, 1], density[idx]

    # Renderizado gráfico
    fig, ax = plt.subplots(figsize=(7, 7))
    sc = ax.scatter(x, y, c=density, cmap='viridis', s=15, alpha=0.8)

    ax.set_xlim(-10, 10)
    ax.set_ylim(-10, 10)
    ax.set_aspect('equal')
    ax.grid(True, linestyle='--', alpha=0.5)

    title_str = (
        fr"$\Sigma = \begin{{bmatrix}} {sigma_11:.2f} & {sigma_12:.2f} \\ {sigma_12:.2f} & {sigma_22:.2f} \end{{bmatrix}}$\n" +
        fr"$\sigma_{{1,1}}={sigma_11:.2f}, \sigma_{{1,2}}={sigma_12:.2f}, \sigma_{{2,2}}={sigma_22:.2f}$"
    )
    ax.set_title(title_str, fontsize=12)
    plt.show()

# Controles interactivos
interact(
    plot_bivariate_covariance,
    sigma_11=widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description=r'$\sigma_{1,1}$'),
    sigma_22=widgets.FloatSlider(value=8.0, min=0.1, max=10.0, step=0.1, description=r'$\sigma_{2,2}$'),
    sigma_12=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1, description=r'$\sigma_{1,2}=\sigma_{2,1}$'))
```
