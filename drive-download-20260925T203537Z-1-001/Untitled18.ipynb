```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal
```

```python
# Definir la media y la matriz de covarianza
mean = np.array([0, 0])
cov = np.array([[10000, 0],
                [0, 10000]])

# Crear una distribución normal multivariada
dist = multivariate_normal(mean=mean, cov=cov)
```

```python
# Crear una malla de puntos para evaluar la PDF
x = np.linspace(-300, 300, 100)
y = np.linspace(-300, 300, 100)
X, Y = np.meshgrid(x, y)
pos = np.empty(X.shape + (2,))
pos[:, :, 0] = X
pos[:, :, 1] = Y

# Calcular la función de densidad de probabilidad (PDF)
Z = dist.pdf(pos)
```

```python
# Dibujar el contorno de la distribución
plt.figure(figsize=(8, 7))
plt.contourf(X, Y, Z, levels=15, cmap='viridis')
plt.colorbar(label='Densidad de probabilidad')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Distribución Normal Bivariada')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
