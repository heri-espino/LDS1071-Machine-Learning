```python
import numpy as np
from scipy.spatial import Delaunay
import plotly.graph_objects as go

# Generación del retículo en N_0^3
x, y, z = np.meshgrid(range(3), range(3), range(3))
points = np.vstack([x.ravel(), y.ravel(), z.ravel()]).T

# Triangulación de Delaunay
tri = Delaunay(points)

# Extracción de las aristas (1-esqueleto del complejo simplicial)
edges = set()
for simplex in tri.simplices:
    for i in range(4):
        for j in range(i+1, 4):
            edges.add(tuple(sorted((simplex[i], simplex[j]))))

# Preparación de vectores para Plotly
edge_x = []
edge_y = []
edge_z = []
for edge in edges:
    p0, p1 = points[edge[0]], points[edge[1]]
    edge_x.extend([p0[0], p1[0], None])
    edge_y.extend([p0[1], p1[1], None])
    edge_z.extend([p0[2], p1[2], None])

# Nodos (0-símplices)
trace_nodes = go.Scatter3d(
    x=points[:,0], y=points[:,1], z=points[:,2],
    mode='markers',
    marker=dict(size=6, color='red', opacity=0.9),
    name='Nodos'
)

# Aristas (1-símplices)
trace_edges = go.Scatter3d(
    x=edge_x, y=edge_y, z=edge_z,
    mode='lines',
    line=dict(color='blue', width=2),
    hoverinfo='none',
    name='Aristas'
)

# Configuración del espacio topológico
layout = go.Layout(
    title="Complejo Simplicial de Delaunay: Fenomenología Kinestésica en R³",
    scene=dict(
        xaxis=dict(title='X', dtick=1, gridcolor='lightgray'),
        yaxis=dict(title='Y', dtick=1, gridcolor='lightgray'),
        zaxis=dict(title='Z', dtick=1, gridcolor='lightgray'),
        bgcolor='white'
    ),
    showlegend=False,
    margin=dict(l=0, r=0, b=0, t=40)
)

fig = go.Figure(data=[trace_edges, trace_nodes], layout=layout)
fig.show()
```
